A Composite Risk Score


Turning 7 individual risk factors into a single point-based score for predicting future customer risk

```python

age_points = {
    "Young Adult": 4, "Early Career": 3, "Mid Life": 2, "Pre-Retirement": 1, "Senior": 0}

interest_points = {
    "Maximum Risk Premium": 3, "High Yield Premium": 2, "Moderate Market Rate": 1, "Low-Cost Financing": 0}

income_points = {
    "Low income": 4, "Lower middle income": 3, "Middle income": 2, "Upper middle income": 1, "High income": 0}

employment_points = {
    "Unemployed": 3, "Part-time": 2, "Self-employed": 1, "Full-time": 0}

cosigner_points = {
    "No": 1, "Yes": 0}

creditscore_points = {
    "poor": 3, "average": 2, "good": 1, "excellent": 0}

dti_points = {
    "Extreme Risk": 3, "High Risk": 2, "Manageable Risk": 1, "Minimal Risk": 0}


df["Age_pts"] = df["Age_label"].map(age_points).astype(float)

df["InterestRate_pts"] = df["InterestRate_label"].map(interest_points).astype(float)

df["Income_pts"] = df["Income_label"].map(income_points).astype(float)

df["Employment_pts"] = df["EmploymentType"].map(employment_points).astype(float)

df["CoSigner_pts"] = df["HasCoSigner"].map(cosigner_points).astype(float)

df["CreditScore_pts"] = df["Creditscore_label"].map(creditscore_points).astype(float)

df["DTI_pts"] = df["DTI_label"].map(dti_points).astype(float)


df["RiskScore"] = (
    df["Age_pts"] + df["InterestRate_pts"] + df["Income_pts"] +
    df["Employment_pts"] + df["CoSigner_pts"] + df["CreditScore_pts"] + df["DTI_pts"])

print(df["RiskScore"].describe())


df["RiskScore"] = (
    df["Age_pts"] + df["InterestRate_pts"] + df["Income_pts"] +
    df["Employment_pts"] + df["CoSigner_pts"] + df["CreditScore_pts"] + df["DTI_pts"])

print(df["RiskScore"].describe())
```

```python
df.head(10)
```

```python
score_default = (df.groupby("RiskScore")["Default"].mean() * 100).round(2)
print(score_default)
```

```python
plt.figure(figsize=(13,6))
ax = sns.barplot(x=score_default.index, y=score_default.values, hue=score_default.index, palette='rocket', legend=False)

plt.title('Actual Default Rate by Composite Risk Score (7-Factor Model)', fontsize=16, fontweight='bold')
plt.xlabel('Risk Score (0 = Lowest Risk, 21 = Highest Risk)')
plt.ylabel('Default Rate (%)')

for i, v in enumerate(score_default.values):
    ax.text(i, v + 0.6, f'{v}%', ha='center', fontsize=9, fontweight='bold')

plt.figtext(0.5, -0.05, "Actual default rate climbs from 0% at score 0 to 59.09% at score 21 — a near-perfect, near-monotonic relationship validating the scoring method.",
            ha='center', fontsize=10, style='italic', color='#3B4A6B')
plt.tight_layout()
plt.show()
```
